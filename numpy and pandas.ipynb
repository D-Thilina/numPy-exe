```python
#pandas retrive csv data set convert into matrix(array) *pd.read_csv(r'path').values*
# r => regular expression

dataset = pd.read_csv(r'C:\Users\thili\OneDrive\Desktop\jupyter note\3-Module\numpy\dataset\student-marks.csv').values

print(dataset)
```

```python
import numpy as np
avg = np.mean(dataset[:,1:], axis = 1)
print(avg)
print("\nsize array")
print(avg.shape)
```

```python
for i in avg:
    print(i)
```

```python
def findGrade(mark):
    if(mark >= 0 and mark < 35):
        grade = 'F'
    elif(mark >= 35 and mark < 45):
        grade = 'S'
    elif(mark >= 45 and mark < 65):
        grade = 'C'
    elif(mark >= 65 and mark < 75):
        grade = 'B'
    elif(mark >= 75 and mark < 100):
        grade = 'A'
    return grade
```

```python
grade = findGrade(95)
print(grade)
```

```python
for i in dataset[:, 1]:
    print(findGrade(i))
```

```python
#genarator object to list
print(list(findGrade(i) for i in dataset[:, 1]))
```

```python
print([findGrade(i) for i in dataset[:, 1]])
```

```python
#convert in np array

array = np.array(list(findGrade(i) for i in dataset[:, 1]))
print(array)
```

```python
#find math grade 

mathsGrade = np.array([findGrade(i) for i in dataset[:,1]])
print(mathsGrade)
mathsGrade = mathsGrade.reshape(-1,1)
print(mathsGrade) #column vector
```

```python
#physics marks

phyGrade = np.array([findGrade(i) for i in dataset[:, 2]])
print(phyGrade)
#conver ==>> vector Physics
phyGrade = phyGrade.reshape(-1,1)
print(phyGrade)
```

```python
#Chemistry Grade

chemGrade = np.array([findGrade(i) for i in dataset[:, 3]])
print(chemGrade)
#convert column vec
chemGrade = chemGrade.reshape(-1,1)
print(chemGrade)
```

```python
avg = avg.reshape(-1,1)
print(avg)
```
